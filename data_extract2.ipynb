```python
#首先更新全A成分股信息
srcdir = "E:/SJTU/intern/gtht/barra/data_base/index_component_日频"
import pandas as pd

df = pd.read_pickle(f"{srcdir}/866011.RI_19_26D_dict.pkl")

temp = rqdatac.index_weights("866011.RI", start_date="2019-01-01", end_date="2020-01-01", market='cn')
trade_dates = temp.index.get_level_values(0).unique()
for dt in trade_dates:
    df[dt] = temp.loc[dt]["weight"]
    print(dt)

import pickle
# 使用内置的 pickle 模块写入文件
with open(f"{srcdir}/866011.RI_19_26D_dict.pkl", "wb") as f:
    pickle.dump(df, f)
```

```python
#其次，提取个股每日收益率
stkdir = "E:/SJTU/intern/gtht/barra/data_base/stk_ret"
df = pd.read_pickle(f"{srcdir}/866011.RI_19_26D_dict.pkl")
res = pd.read_pickle(f"{stkdir}/全A_ret_20_26D_dict.pkl")

for dt in list(df.keys()):
    if dt in list(res.keys()):
        continue
    stk = df[dt].index.tolist()
    stk_fb = [s for s in stk if not s.endswith(".BJSE")]
    df_ret = get_price_change_rate(stk_fb, start_date=dt, end_date=dt, expect_df=True, market='cn')
    res[dt] = df_ret.T[dt]
    res[dt].name = "ret"
    print(dt)
    
with open(f"{stkdir}/全A_ret_20_26D_dict.pkl", "wb") as f:
    pickle.dump(res, f)
res
```
